```python
games = duckdb.sql("""
    SELECT app_id, name, release_date, price, tags, recommendations, genres, categories, median_playtime_forever, median_playtime_2weeks, user_score, peak_ccu, publishers
    FROM read_csv_auto('D:/OneDrive - kuhn edv-Dienstleistungen/Data Science/semester_2/info/MV/HAusarbeit/steam_games.csv', ignore_errors=true)
""").df()
```

```python
joined = duckdb.sql("""SELECT *
FROM reviews
LEFT JOIN games
ON reviews.app_id = games.app_id """)

print(joined)
```
